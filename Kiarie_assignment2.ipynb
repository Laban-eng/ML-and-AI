```python
# Function to perform the conversion and check the threshold
def process_temperature(celsius, threshold=30.0):
    """Converts Celsius to Fahrenheit and checks against a threshold."""

    # 1. Convert C to F
    fahrenheit = (celsius * 9/5) + 32
    print(f"Result: {celsius}°C is exactly {fahrenheit}°F")

    # 2. Report if it is above or below the threshold
    if celsius > threshold:
        print(f"Alert: This is ABOVE our safety threshold of {threshold}°C!\n")
    elif celsius < threshold:
        print(f"Status: This is below the threshold of {threshold}°C.\n")
    else:
        print(f"Status: This is exactly at the threshold of {threshold}°C.\n")

# --- Main Program Loop ---
print("--- Temperature Conversion Factory ---")
print("Type 'quit' whenever you want to stop the machine.\n")

# 3. Loop to convert several temperatures
while True:
    # 4. Ask the user to enter a temperature
    user_input = input("Enter a temperature in °C: ")

    # Check if the user wants to stop the loop
    if user_input.lower() == 'quit':
        print("Powering down the factory. Goodbye!")
        break

    # Safety check to ensure they typed a number, not random letters
    try:
        celsius_value = float(user_input)
        # Pass the user's number into our custom machine (the function)
        process_temperature(celsius_value)
    except ValueError:
        print("Error: Please enter a valid number.\n")
```
